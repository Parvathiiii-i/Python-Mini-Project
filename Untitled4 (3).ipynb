```python
from datetime import datetime, timedelta
import json
import os

# Custom Exceptions for clean domain error handling
class LibraryException(Exception):
    """Base exception class for library domain errors."""
    pass

class BookNotFoundError(LibraryException):
    pass

class BookUnavailableError(LibraryException):
    pass

class MemberNotFoundError(LibraryException):
    pass


class Book:
    """Represents an individual book in the library collection."""

    def __init__(self, book_id, title, author, category, copies=1):
        self.book_id = str(book_id)
        self.title = title.title().strip()
        self.author = author.title().strip()
        self.category = category.title().strip()
        self.total_copies = int(copies)
        self.available_copies = int(copies)

    def to_dict(self):
        """Serialize instance to dictionary for storage."""
        return {
            "book_id": self.book_id,
            "title": self.title,
            "author": self.author,
            "category": self.category,
            "total_copies": self.total_copies,
            "available_copies": self.available_copies
        }

    @classmethod
    from_dict = lambda cls, data: cls._from_dict_impl(data)

    @classmethod
    def _from_dict_impl(cls, data):
        book = cls(
            data["book_id"],
            data["title"],
            data["author"],
            data["category"],
            data["total_copies"]
        )
        book.available_copies = data["available_copies"]
        return book


class Member:
    """Represents a registered library member."""

    def __init__(self, member_id, name, email):
        self.member_id = str(member_id)
        self.name = name.title().strip()
        self.email = email.lower().strip()
        # Dictionary storing borrowed books: {book_id: due_date_str}
        self.borrowed_books = {}

    def to_dict(self):
        """Serialize instance to dictionary for storage."""
        return {
            "member_id": self.member_id,
            "name": self.name,
            "email": self.email,
            "borrowed_books": self.borrowed_books
        }

    @classmethod
    def from_dict(cls, data):
        member = cls(data["member_id"], data["name"], data["email"])
        member.borrowed_books = data.get("borrowed_books", {})
        return member


class LibraryService:
    """Encapsulates core business logic, state management, and file persistence."""

    DATA_FILE = "library_data.json"
    DAILY_FINE_RATE = 2.0  # $2 per overdue day

    def __init__(self):
        # Data structures: Dictionaries for O(1) lookups, Sets for quick filtering
        self.books = {}      # {book_id: Book instance}
        self.members = {}    # {member_id: Member instance}
        self.categories = set()
        self._load_data()

    def _load_data(self):
        """Loads records from JSON file with file handling exception safety."""
        if not os.path.exists(self.DATA_FILE):
            self._seed_default_data()
            return

        try:
            with open(self.DATA_FILE, "r", encoding="utf-8") as f:
                data = json.load(f)

                for b_data in data.get("books", []):
                    book = Book.from_dict(b_data)
                    self.books[book.book_id] = book
                    self.categories.add(book.category)

                for m_data in data.get("members", []):
                    member = Member.from_dict(m_data)
                    self.members[member.member_id] = member

        except (json.JSONDecodeError, IOError) as e:
            print(f"\n[Warning] Failed to load data from file: {e}. Starting fresh.")

    def save_data(self):
        """Persists catalog and member records to JSON file."""
        data = {
            "books": [b.to_dict() for b in self.books.values()],
            "members": [m.to_dict() for m in self.members.values()]
        }
        try:
            with open(self.DATA_FILE, "w", encoding="utf-8") as f:
                json.dump(data, f, indent=4)
        except IOError as e:
            print(f"\n[Error] Unable to save data to file: {e}")

    def _seed_default_data(self):
        """Initializes system with sample data if no storage file exists."""
        default_books = [
            Book("B101", "Clean Code", "Robert C. Martin", "Programming", 3),
            Book("B102", "The Pragmatic Programmer", "Andrew Hunt", "Programming", 2),
            Book("B103", "Dune", "Frank Herbert", "Sci-Fi", 4),
            Book("B104", "1984", "George Orwell", "Fiction", 2)
        ]
        default_members = [
            Member("M101", "Alice Smith", "alice@example.com"),
            Member("M102", "Bob Jones", "bob@example.com")
        ]

        for b in default_books:
            self.books[b.book_id] = b
            self.categories.add(b.category)

        for m in default_members:
            self.members[m.member_id] = m

        self.save_data()

    # --- BOOK MANAGEMENT ---
    def add_book(self, book_id, title, author, category, copies):
        if book_id in self.books:
            # If book exists, increment available & total copies
            existing = self.books[book_id]
            existing.total_copies += copies
            existing.available_copies += copies
        else:
            new_book = Book(book_id, title, author, category, copies)
            self.books[book_id] = new_book
            self.categories.add(new_book.category)

        self.save_data()

    def search_books(self, query):
        """Searches books by title, author, or category."""
        query = query.lower().strip()
        results = [
            b for b in self.books.values()
            if query in b.title.lower() or query in b.author.lower() or query in b.category.lower()
        ]
        return results

    # --- MEMBER MANAGEMENT ---
    def add_member(self, member_id, name, email):
        if member_id in self.members:
            raise LibraryException(f"Member ID '{member_id}' is already registered.")

        new_member = Member(member_id, name, email)
        self.members[member_id] = new_member
        self.save_data()

    # --- CIRCULATION LOGIC ---
    def issue_book(self, book_id, member_id, days=14):
        if book_id not in self.books:
            raise BookNotFoundError(f"Book ID '{book_id}' not found.")
        if member_id not in self.members:
            raise MemberNotFoundError(f"Member ID '{member_id}' not found.")

        book = self.books[book_id]
        member = self.members[member_id]

        if book.available_copies <= 0:
            raise BookUnavailableError(f"'{book.title}' has no copies currently available.")
        if book_id in member.borrowed_books:
            raise LibraryException(f"Member '{member.name}' has already borrowed this book.")

        # Process checkout
        book.available_copies -= 1
        due_date = (datetime.now() + timedelta(days=days)).strftime("%Y-%m-%d")
        member.borrowed_books[book_id] = due_date

        self.save_data()
        return book.title, due_date

    def return_book(self, book_id, member_id):
        if book_id not in self.books:
            raise BookNotFoundError(f"Book ID '{book_id}' not found.")
        if member_id not in self.members:
            raise MemberNotFoundError(f"Member ID '{member_id}' not found.")

        book = self.books[book_id]
        member = self.members[member_id]

        if book_id not in member.borrowed_books:
            raise LibraryException(f"Member '{member.name}' does not have this book checked out.")

        due_date_str = member.borrowed_books.pop(book_id)
        book.available_copies += 1

        # Fine Calculation
        due_date = datetime.strptime(due_date_str, "%Y-%m-%d")
        days_overdue = (datetime.now() - due_date).days
        fine = max(0.0, days_overdue * self.DAILY_FINE_RATE)

        self.save_data()
        return book.title, fine


class ConsoleUI:
    """Handles terminal interaction, menu rendering, and input validation."""

    def __init__(self):
        self.service = LibraryService()

    @staticmethod
    def get_non_empty_string(prompt):
        while True:
            val = input(prompt).strip()
            if val:
                return val
            print("Input cannot be empty. Please try again.")

    @staticmethod
    def get_positive_int(prompt):
        while True:
            try:
                val = int(input(prompt))
                if val > 0:
                    return val
                print("Value must be greater than zero.")
            except ValueError:
                print("Invalid numerical input. Please enter an integer.")

    def display_menu(self):
        print("\n" + "="*45)
        print("    SMART LIBRARY MANAGEMENT SYSTEM")
        print("="*45)
        print("1. View All Books")
        print("2. Search Books")
        print("3. Add New Book")
        print("4. Register New Member")
        print("5. View All Members")
        print("6. Issue Book")
        print("7. Return Book")
        print("8. Exit Application")
        print("="*45)

    def run(self):
        while True:
            self.display_menu()
            choice = input("Enter choice (1-8): ").strip()

            if choice == "1":
                self._view_books()
            elif choice == "2":
                self._search_books()
            elif choice == "3":
                self._add_book()
            elif choice == "4":
                self._add_member()
            elif choice == "5":
                self._view_members()
            elif choice == "6":
                self._issue_book()
            elif choice == "7":
                self._return_book()
            elif choice == "8":
                print("\nThank you for using Smart Library System. Goodbye!")
                break
            else:
                print("Invalid selection! Please enter a digit between 1 and 8.")

    def _view_books(self):
        print("\n--- CATALOG OVERVIEW ---")
        if not self.service.books:
            print("No books currently in the catalog.")
            return

        print(f"{'ID':<8} | {'Title':<30} | {'Author':<20} | {'Category':<15} | {'Avail/Total'}")
        print("-" * 88)
        for b in self.service.books.values():
            status = f"{b.available_copies}/{b.total_copies}"
            print(f"{b.book_id:<8} | {b.title[:28]:<30} | {b.author[:18]:<20} | {b.category[:13]:<15} | {status}")

    def _search_books(self):
        print("\n--- SEARCH CATALOG ---")
        query = self.get_non_empty_string("Enter search term (Title/Author/Category): ")
        results = self.service.search_books(query)

        if not results:
            print(f"No books matched search term: '{query}'")
            return

        print(f"\nFound {len(results)} match(es):")
        print(f"{'ID':<8} | {'Title':<30} | {'Author':<20} | {'Available'}")
        print("-" * 70)
        for b in results:
            print(f"{b.book_id:<8} | {b.title[:28]:<30} | {b.author[:18]:<20} | {b.available_copies}")

    def _add_book(self):
        print("\n--- ADD BOOK TO CATALOG ---")
        book_id = self.get_non_empty_string("Enter Book ID (e.g. B105): ").upper()
        title = self.get_non_empty_string("Enter Title: ")
        author = self.get_non_empty_string("Enter Author: ")
        category = self.get_non_empty_string("Enter Category: ")
        copies = self.get_positive_int("Enter Number of Copies: ")

        self.service.add_book(book_id, title, author, category, copies)
        print(f"Success: '{title}' added/updated in the catalog.")

    def _add_member(self):
        print("\n--- MEMBER REGISTRATION ---")
        member_id = self.get_non_empty_string("Enter Member ID (e.g. M103): ").upper()
        name = self.get_non_empty_string("Enter Full Name: ")
        email = self.get_non_empty_string("Enter Email Address: ")

        try:
            self.service.add_member(member_id, name, email)
            print(f"Success: Member '{name}' registered successfully.")
        except LibraryException as e:
            print(f"Error: {e}")

    def _view_members(self):
        print("\n--- REGISTERED MEMBERS ---")
        if not self.service.members:
            print("No registered members found.")
            return

        print(f"{'ID':<8} | {'Name':<20} | {'Email':<25} | {'Active Loans'}")
        print("-" * 70)
        for m in self.service.members.values():
            loans = len(m.borrowed_books)
            print(f"{m.member_id:<8} | {m.name[:18]:<20} | {m.email[:23]:<25} | {loans}")

    def _issue_book(self):
        print("\n--- CHECKOUT BOOK ---")
        book_id = self.get_non_empty_string("Enter Book ID: ").upper()
        member_id = self.get_non_empty_string("Enter Member ID: ").upper()

        try:
            title, due_date = self.service.issue_book(book_id, member_id)
            print(f"\nSuccess: '{title}' issued successfully.")
            print(f"Return Due Date: {due_date}")
        except LibraryException as e:
            print(f"Issue Failed: {e}")

    def _return_book(self):
        print("\n--- RETURN BOOK ---")
        book_id = self.get_non_empty_string("Enter Book ID: ").upper()
        member_id = self.get_non_empty_string("Enter Member ID: ").upper()

        try:
            title, fine = self.service.return_book(book_id, member_id)
            print(f"\nSuccess: '{title}' returned successfully.")
            if fine > 0:
                print(f"Overdue Fine Assessed: ${fine:.2f}")
            else:
                print("Returned on time. Zero fine assessed.")
        except LibraryException as e:
            print(f"Return Failed: {e}")


if __name__ == "__main__":
    app = ConsoleUI()
    app.run()
```
